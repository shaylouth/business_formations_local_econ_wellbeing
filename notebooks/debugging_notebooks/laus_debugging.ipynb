# MISSING MONTHS

```python
missing_months = laus.loc[(laus['months_observed']!= 12)]
print(missing_months['state'].unique())
print(missing_months['year'].unique())
```

```python

month_count = missing_months['months_observed'].unique()

missing_by_month_count = {}

for count in month_count:
    missing_by_month_count[count] = missing_months.loc[(missing_months['months_observed'] == count)]
    print(f'Count: {count}')
    print(missing_by_month_count[count]['state'].unique())
    print('#####################')
```

```python
for count, df in missing_by_month_count.items():
    print(f"Count: {count}")
    print(df['year'].value_counts().sort_index())
    print("#####################")

missing_months.to_csv('missing_months.csv')
```
